#### 🧹 Group-based age missing values imputation

````python
age_median_global=train_df['Age'].median()
age_median_map=train_df.groupby(['Pclass','FamilySize'])['Age'].median()

def fill_age(row):
    if pd.isnull(row['Age']):
        key=(row['Pclass'],row['FamilySize'])

        if key in age_median_map.index:
            #判断元祖key是否在map里，index用来展开所有的排列
            age=age_median_map.loc[key]
            #loc(location) 根据 key 元祖tuple定位到对应的median
            if pd.notnull(age):
                return age
        return age_median_global
    return row['Age']

train_df['Age']=train_df.apply(fill_age,axis=1)
#axis=1 相当于按行操作，类似于调用函数fill_age 里面的for循环装填row 


test_df['Age']=test_df.apply(fill_age,axis=1)
#因为在定义函数fill_age里面age_median_map是按照train_df返回的,这里只是fill_age函数应用到test_df里并返回到test_df里
````

````python
print(train_df['Age'].isnull().sum())
print(test_df['Age'].isnull().sum())
````

<div style="background:#FFF8E6;padding:15px;border-left:6px solid #F39C12;border-radius:8px;">
    when we find <b>NaN values in one column</b>, we use <b>train_df.isnull().sum(), while <b>search NaN value in one row with specific feature</b>, we use <b>pd.isnull(row['Age'])</b>

### 💻 Feature Selection

````python
baseline_features=['Pclass','Sex','Age','Fare','Embarked','Deck','FamilySize']
````

````python
import sklearn.model_selection
import sklearn.ensemble
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import RandomForestClassifier

def evaluate_features(features):
    X=train_df[features].copy()
    X=pd.get_dummies(
        X,
        drop_first=True
    )
    y=train_df['Survived']

    model=RandomForestClassifier(
        n_estimators=300,
        max_depth=4,
        min_samples_leaf=3,
        random_state=42,
        n_jobs=-1
    )
    
    cv=StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=42
    )
        
    scores=cross_val_score(
        model,
        X,
        y,
        cv=cv,
        scoring='accuracy'
    )
    print(f"features: {features}")
    print(f"--------------------")
    for score in scores:
        print(f"{score:.2f}")
    print(f"--------------------")
    print(f"{scores.mean():.4f}")
    return scores.mean()
````

````python
baseline_score=evaluate_features(baseline_features)
````

````python
title_score=evaluate_features(baseline_features+['Title'])
````

````python
IsAlone_score=evaluate_features(baseline_features+['Title']+['IsAlone'])
````

````python
FarePerPerson1_score=evaluate_features(baseline_features+['Title']+['IsAlone']+['FarePerPerson'])
````

````python
FarePerPerson2_score=evaluate_features(baseline_features+['Title']+['FarePerPerson'])
````

````python
cv_results=pd.DataFrame({"Feature Set": ['Baseline','+Title','+Title+IsAlone','+Title+IsAlone+FarePerPerson','+Title+FarePerPerson'],
                         "Mean CV":[baseline_score,title_score,IsAlone_score,FarePerPerson1_score,FarePerPerson2_score]})
cv_results
````

### 💻 Hyperparameter tuning

````python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score, StratifiedKFold

cv=StratifiedKFold(
    n_splits=5,
    random_state=42,
    shuffle=True
)
def evaluate_model(
    X,
    y,
    n_estimators=100,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features='sqrt' 
):
    model=RandomForestClassifier(
        random_state=42,
        n_estimators=n_estimators,
        max_depth=max_depth,
        min_samples_split=min_samples_split,
        min_samples_leaf=min_samples_leaf,
        max_features=max_features,
        n_jobs=-1,
    )
    scores=cross_val_score(
        model,
        X,
        y,
        cv=cv,
        scoring='accuracy'
    )

    for score in scores:
        print(f'{score:.4f}')
    print(f'------------------')
    print(f'{scores.mean():.4f}')
````
